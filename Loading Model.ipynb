```python
import joblib

# Load the saved model
loaded_model = joblib.load('landslide_prediction_model.pkl')

print("Model loaded successfully!")
```

```python
import numpy as np

# New data sample (Latitude, Longitude, Month, Week, Rainfall_1Day, Rainfall_3Days, Rainfall_7Days)
new_data = np.array([[45.678, 90.123, 5, 3, 78.2, 200.5, 450.0]])  # Example values

# Make prediction
prediction = loaded_model.predict(new_data)

# Show result
if prediction[0] == 1:
    print("Warning: Landslide Risk Detected!")
else:
    print("No Landslide Risk.")
```
